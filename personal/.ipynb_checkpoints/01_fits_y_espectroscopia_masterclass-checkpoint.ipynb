### 🪐 ¿Qué es un Disco Protoplanetario?
Imagina un **gigantesco plato giratorio de gas (99%) y polvo (1%)** en cuyo centro acaba de nacer una estrella (un "sol bebé"). 
- El material de este plato no está estático: está girando a miles de kilómetros por hora impulsado por la gravedad de la estrella.
- Dentro de este plato de gas y polvo, la materia empieza a agruparse por gravedad. **Aquí es exactamente donde se están formando nuevos planetas (exoplanetas)** como la Tierra o Júpiter.

## 🌈 2. ¿Qué es un Espectro? La Analogía del "Código de Barras de Luz"

### 💡 Luz y Longitud de Onda ($\lambda$)
La luz viaja en forma de ondas. La distancia entre dos crestas consecutivas de la onda se llama **longitud de onda** ($\lambda$, la letra griega *lambda*).
- **Luz Visible:** Lo que ven nuestros ojos varía entre $0.4\,\mu	ext{m}$ (luz violeta) y $0.7\,\mu	ext{m}$ (luz roja).
- **Micrómetro ($\mu	ext{m}$):** Es una millonésima parte de un metro ($1\,\mu	ext{m} = 0.000001	ext{ metros}$).
- **Infrarrojo:** El gas de Monóxido de Carbono (CO) que estudiaremos en el disco emite luz en el **infrarrojo cercano ($pprox 4.7\,\mu	ext{m}$)**. Esta luz es invisible para el ojo humano, pero puede ser captada por grandes telescopios infrarrojos en la Tierra (como el VLT en Chile o Gemini en Hawaii).

### 🏷️ El Código de Barras Químico
Cuando la luz de un disco protoplanetario pasa a través de un telescopio equipado con un **espectrógrafo** (un instrumento que actúa como un prisma súper potente), la luz se separa en todos sus "colores" (longitudes de onda).

Cada molécula (como el **Monóxido de Carbono, CO**) actúa como un **código de barras único**: absorbe o emite luz en longitudes de onda muy específicas. Si vemos los picos de ese código de barras a $4.7\,\mu	ext{m}$, **sabemos con 100% de certeza que hay gas de CO en el disco**.

## 📦 3. ¿Qué es un archivo FITS y por qué no usamos JPG o PNG?

Cuando tomas una foto con tu celular, se guarda como un archivo `.jpg` o `.png`. Esos formatos **comprimen la imagen** y alteran la luz real para que se vea bonita en una pantalla. 

Para hacer ciencia, la NASA y la Unión Astronómica Internacional (IAU) inventaron el formato **FITS** (*Flexible Image Transport System*).

### 🧰 Anatomía de un archivo FITS: La "Caja con Bitácora"
Un archivo FITS es como una caja que contiene dos cosas esenciales:

1. **El Encabezado (*Header*):** Una bitácora de texto que registra exactamente las condiciones de la observación:
   - Nombre de la estrella u objeto (ej. `TARGNAME = 'AA Tau'`).
   - Telescopio u instrumento utilizado (ej. `TELESCOP = 'ESO-VLT'`).
   - Coordenadas en el cielo, fecha, tiempo de exposición y calibraciones.
   - **Parámetros del Sistema de Coordenadas del Mundo (WCS):** Las claves que nos dicen qué unidades tienen los ejes.

2. **Los Datos (*Data*):** Los números científicos **puros, sin comprimir**, leídos directamente por el detector digital del telescopio.

3. **HDUs (*Header Data Units*):** Un archivo FITS puede tener varias "páginas" o secciones. Cada sección se llama **HDU**:
   - `PrimaryHDU`: La portada y datos principales.
   - `BinTableHDU`: Una **tabla binaria** súper eficiente que guarda columnas de números (por ejemplo: columna 1 = Longitud de onda, columna 2 = Flujo de luz, columna 3 = Error).

## 🛠️ 4. Manos a la Obra: Carga y Exploración con Python

Cargaremos las librerías estándar de astronomía explicando qué hace cada una:


```python
# astropy.io.fits: La herramienta oficial para abrir y manipular archivos astronómicos FITS
from astropy.io import fits

# numpy: La librería matemática para manejar arreglos de números y vectores de alta velocidad
import numpy as np

# matplotlib.pyplot: Nuestra herramienta para dibujar gráficos científicos de alta resolución
import matplotlib.pyplot as plt

# pandas: Para manejar tablas de datos (como hojas de cálculo de Excel) de manera profesional
import pandas as pd

# Configurar el tamaño y calidad visual de los gráficos
plt.rcParams['figure.dpi'] = 140
plt.rcParams['font.size'] = 11
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

print("✅ Librerías astronómicas cargadas con éxito.")
```

---
## 🔍 5. Abrir e Inspeccionar un Archivo FITS Real (`AATau_M.fits`)

El objeto **AA Tau** es una estrella joven (tipo T Tauri) rodeada por un disco protoplanetario ubicado a unos 450 años luz de la Tierra en la constelación de Tauro.

Abramos su archivo FITS ubicado en `../data/discos/AATau_M.fits`.


```python
# Ruta al archivo del disco AA Tau
fits_filename = '../data/discos/AATau_M.fits'

# Abrimos la "caja FITS" usando fits.open()
with fits.open(fits_filename) as hdul:
    print(f"📁 Nombre de archivo: {fits_filename}")
    print(f"📄 Número de secciones/páginas (HDUs): {len(hdul)}\n")
    
    # hdul.info() imprime un resumen de lo que contiene cada HDU
    print("--- ESTRUCTURA INTERNA DEL ARCHIVO FITS ---")
    hdul.info()
```

### 💡 ¿Qué nos dice la salida anterior?
- **HDU 0 (`PRIMARY`):** Es la portada (*PrimaryHDU*). Contiene 547 líneas de metadatos sobre la observación del telescopio.
- **HDU 1 (`BinTableHDU`):** Es una **Tabla Binaria** (*BinTableHDU*) que contiene 1 fila con 3 columnas de datos (`[12860D, 12860D, 12860D]`). Significa que hay **12,860 puntos de medición espectral** en esa tabla.

Inspeccionemos los nombres de esas 3 columnas de datos:


```python
with fits.open(fits_filename) as hdul:
    # Seleccionamos la extensión HDU 1 que contiene la tabla binaria
    tabla_hdu = hdul[1]
    
    # Imprimimos las columnas de la tabla binaria
    print("--- COLUMNAS DE LA TABLA BINARIA ---")
    for i, col in enumerate(tabla_hdu.columns):
        print(f"Columna {i+1}: Nombre = {col.name:<15} | Formato = {col.format:<8} | Unidad = {col.unit}")
```

---
## 📊 6. Extraer los Datos Físicos: Longitud de Onda, Flujo y Error

Ahora extraeremos los 3 arreglos numéricos clave:
1. **`WAVELENGTH` (Longitud de Onda $\lambda$):** La posición de la luz en el espectro, medida en micrómetros ($\mu	ext{m}$). Es el eje horizontal ($X$).
2. **`FLUX` (Flujo de Luz):** La cantidad de energía de luz que captó el detector del telescopio. Es el eje vertical ($Y$).
3. **`UNCERTAINTY` (Incertidumbre / Error $\sigma$):** El margen de error o ruido experimental asociado a cada punto de medición.


```python
with fits.open(fits_filename) as hdul:
    data_table = hdul[1].data
    
    # Extraemos los arreglos numéricos (accediendo a la primera fila [0])
    longitud_onda = data_table['WAVELENGTH'][0]
    flujo         = data_table['FLUX'][0]
    error_medicion = data_table['UNCERTAINTY'][0]

print(f"Total de mediciones espectrales: {len(longitud_onda)}")
print(f"Longitud de onda mínima: {longitud_onda.min():.4f} μm")
print(f"Longitud de onda máxima: {longitud_onda.max():.4f} μm")
```

---
## 📈 7. Graficar el Espectro Completo del Disco AA Tau

Graficaremos el espectro completo. Observa la forma general:
- El **continuo** es la línea base o el "piso" del gráfico (la luz térmica proveniente de la estrella central y del polvo caliente del disco).
- Las **líneas de emisión** son los "picos o montañas" que sobresalen por encima de la línea base. ¡Esos picos son producidos por las moléculas de gas de Monóxido de Carbono (CO)!


```python
fig, ax = plt.subplots(figsize=(12, 5))

# Dibujamos el espectro: X = longitud de onda, Y = flujo
ax.plot(longitud_onda, flujo, color='#1f77b4', linewidth=0.8, label='Espectro AA Tau (VLT/CRIRES)')

ax.set_xlabel('Longitud de Onda [μm]  ──►  (Color / Frecuencia de la luz)', fontsize=12)
ax.set_ylabel('Flujo [W / m² / μm]  ──►  (Brillo / Energía recibida)', fontsize=12)
ax.set_title('Espectro Infrarrojo Cercano (~4.7 μm) del Disco Protoplanetario AA Tau', fontsize=13, fontweight='bold')
ax.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

## 🧬 8. Entendiendo la Física Molecular y el Catálogo HITRAN

### ⚛️ ¿Por qué la molécula de Monóxido de Carbono (CO) emite luz?
Una molécula de **CO** está formada por un átomo de Carbono (C) y uno de Oxígeno (O) unidos por un enlace químico. Imagínalos como dos esferas unidas por un resorte:
1. **Vibración:** El resorte se estira y se comprime (cambio de nivel vibracional $v$).
2. **Rotación:** Las dos esferas giran una alrededor de la otra (cambio de nivel rotacional $J$).

Cuando el gas caliente del disco protoplanetario sufre colisiones, la molécula de CO se excita a un nivel de energía más alto. Al volver a un nivel más bajo, **libera la energía sobrante emitiendo un fotón de luz** a una frecuencia exacta.

- **Transiciones de la banda fundamental $v=1 
ightarrow 0$:** Ocurren a longitudes de onda de $pprox 4.6 - 4.9\,\mu	ext{m}$.
- **Rama P ($\Delta J = -1$):** Ocurre a longitudes de onda un poco mayores.
- **Rama R ($\Delta J = +1$):** Ocurre a longitudes de onda un poco menores.

### 📖 El Catálogo HITRAN (`data/HITRAN_CO_lineas.xlsx`)
Físicos cuánticos han medido en laboratorio las posiciones exactas de todas las líneas de emisión del CO. Esos datos están guardados en la tabla `HITRAN_CO_lineas.xlsx`.

Carguemos el catálogo con `pandas`:


```python
# Cargamos la hoja 'COlineranges' del archivo Excel HITRAN
hitran_df = pd.read_excel('../data/HITRAN_CO_lineas.xlsx', sheet_name='COlineranges')

print("--- CATÁLOGO HITRAN DE LÍNEAS DE MONÓXIDO DE CARBONO (CO) ---")
print(hitran_df.head(12))
```

## 🎯 9. Zoom Científico: Identificando la Línea $P(1)$ a $4.674151\,\mu	ext{m}$

En la tabla HITRAN leemos que la línea llamada **$P(1)$** ocurre exactamente a una longitud de onda teórica de:
$$\lambda_{teórica} = 4.674151\,\mu	ext{m}$$

Hagamos un "zoom" en nuestro gráfico dentro de una pequeña ventana espectral alrededor de esa longitud de onda ($\pm 0.005\,\mu	ext{m}$) para ver cómo luce la línea de emisión real captada por el telescopio.


```python
lambda_p1 = 4.674151 # Longitud de onda teórica de P(1) en μm

# Creamos una "máscara" o filtro booleano para seleccionar solo los puntos cercanos a P(1)
ventana = (longitud_onda >= lambda_p1 - 0.006) & (longitud_onda <= lambda_p1 + 0.006)

# Extraemos las sub-ventanas de datos
wl_zoom = longitud_onda[ventana]
flujo_zoom = flujo[ventana]
error_zoom = error_medicion[ventana]

fig, ax = plt.subplots(figsize=(10, 5))

# Graficamos la línea medida por el telescopio
ax.plot(wl_zoom, flujo_zoom, color='black', linewidth=1.5, marker='o', markersize=3, label='Espectro Medido (AA Tau)')

# Dibujamos una línea punteada roja en la posición exacta teórica de HITRAN
ax.axvline(lambda_p1, color='red', linestyle='--', linewidth=1.5, label=f'Línea P(1) HITRAN ({lambda_p1:.6f} μm)')

# Sombreado de la incertidumbre/error de medición
ax.fill_between(wl_zoom, flujo_zoom - error_zoom, flujo_zoom + error_zoom, color='gray', alpha=0.3, label='Margen de Error (σ)')

ax.set_xlabel('Longitud de Onda [μm]', fontsize=12)
ax.set_ylabel('Flujo [W / m² / μm]', fontsize=12)
ax.set_title('Zoom en la Línea de Emisión CO v1-0 P(1) en el Disco AA Tau', fontsize=13, fontweight='bold')
ax.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

### 🔑 Puntos Clave Aprendidos:
1. **Discos Protoplanetarios:** Son estructuras giratorias de gas y polvo alrededor de estrellas recién nacidas donde nacen los planetas.
2. **Espectro Astronómico:** Es el "código de barras de luz" que se obtiene al separar la luz en sus longitudes de onda ($\lambda$). Nos dice qué química y qué temperatura hay en el espacio.
3. **Archivos FITS:** Son el estándar de la NASA y la IAU que guardan datos científicos puros en tablas binarias (`BinTableHDU`) junto a una bitácora de metadatos (`Header`).
4. **Líneas de CO y HITRAN:** El gas de CO se excita térmicamente en el disco y emite picos de luz en el infrarrojo ($pprox 4.7\,\mu	ext{m}$). Comparamos los datos del telescopio con el catálogo de laboratorio **HITRAN**.
